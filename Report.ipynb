# Predictive Maintenance in Solar Power System using Machine Learning

## Introduction

Solar energy is one of the most important renewable energy sources. However, solar panels require regular maintenance to ensure optimal performance.

Manual inspection of solar systems is time-consuming and costly. Therefore, this project focuses on building a machine learning-based system that can automatically detect when maintenance is required.

## Problem Statement

The main objective of this project is to predict solar power generation using environmental and sensor data, and identify maintenance requirements based on the difference between predicted and actual power output.

If the deviation between predicted and actual power is high, the system flags a maintenance alert.

## Dataset Description

The dataset used in this project consists of two main parts:

1. Generation Data:
   - AC Power
   - DC Power
   - Daily Yield
   - Total Yield

2. Weather Sensor Data:
   - Irradiation
   - Ambient Temperature
   - Module Temperature

Both datasets were merged using DATE_TIME and PLANT_ID to create a unified dataset.

## Data Preprocessing

The following preprocessing steps were performed:

- Converted DATE_TIME columns to proper datetime format
- Merged generation and weather datasets
- Removed missing values using dropna()
- Filtered out rows where IRRADIATION = 0 (night-time data)
- Selected relevant features for model training

## Feature Selection

The input features used for training the model are:

- Irradiation
- Ambient Temperature
- Module Temperature

The target variable is:

- AC Power

## Machine Learning Model

A Random Forest Regressor was used to predict solar power output.

Reasons for choosing Random Forest:
- Handles non-linear relationships
- Provides high accuracy
- Robust to noise in data

## Neural Network Model

In addition to the Random Forest Regressor, a feed-forward neural network was implemented to evaluate its performance on solar power prediction.

### Architecture
- Input Layer: 3 neurons (Irradiation, Ambient Temperature, Module Temperature)
- Hidden Layers: Two fully connected layers with 64 and 32 neurons respectively, using ReLU activation
- Output Layer: 1 neuron predicting AC Power

### Training Details
- Optimizer: Adam
- Loss Function: Mean Squared Error (MSE)
- Epochs: 100
- Batch Size: 32

### Results
The neural network achieved an R² score of approximately 0.95, slightly lower than the Random Forest model but still demonstrating strong predictive capability.

### Insights
- Neural networks can capture complex non-linear relationships in the data.
- Performance can be further improved with hyperparameter tuning and larger datasets.
- Suitable for future integration with real-time streaming data and deep learning approaches such as LSTM for time-series forecasting.

## Maintenance Detection Logic

The system compares predicted and actual power values.

If the deviation exceeds 20%, the system flags a maintenance alert.

Condition used:

If |Actual - Predicted| / Actual > 0.2 → Maintenance Required
Else → Working Fine

## Results

The model achieved a high R² score of approximately 0.97, indicating excellent prediction accuracy.

Most predicted values closely match the actual values, showing that the model performs well on the dataset.
